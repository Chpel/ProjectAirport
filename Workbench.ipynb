```python
import matplotlib.pyplot as plt

#x10, y10, x11, y11 = list(map(float, input().split(' ')))
#x20, y20, x21, y21 = list(map(float, input().split(' ')))

x10, y10, x11, y11 = 3, 2, 2, 3
x20, y20, x21, y21 = 2, 2, 3, 3

msg = 'ok'
if (x11 == x21) and (y11 == y21):
    msg = 'Пересечение в конце'
elif (x11 == x21) and (x10 == x20) and (y11 == y20) and (y21 == y10):
    msg = 'Пересечение посередине пути (по горизонтали)'
elif (y11 == y21) and (y10 == y20) and (x11 == x20) and (x21 == x10):
    msg = 'Пересечение посередине пути (по вертикали)'
elif (x10 == x21) and (x11 == x20) and (x10 == x21) and (x11 == x20):
    msg = 'Встречное движение'



print(msg)
plt.arrow(x10, y10, x11-x10, y11-y10, length_includes_head=True, head_width = 0.1)
plt.arrow(x20, y20, x21-x20, y21-y20, length_includes_head=True, head_width = 0.1)
plt.show()
```
